```python
#percentage of unavaliable

import pandas as pd

calendar = pd.read_csv("calendar.csv.gz")
listings = pd.read_csv("listings.csv.gz")

# Convert dates into actual dates
calendar["date"] = pd.to_datetime(calendar["date"])

# Convert t/f into True/False
calendar["available"] = calendar["available"].map({
    "t": True,
    "f": False
})

# Attach neighbourhood information to each calendar row
calendar = calendar.merge(
    listings[["id", "neighbourhood_cleansed"]],
    left_on="listing_id",
    right_on="id"
)

# Calculate unavailable percentage by neighbourhood
unavailable = calendar.groupby(
    "neighbourhood_cleansed"
)["available"].apply(
    lambda x: (~x).mean() * 100
)

print(unavailable)
```

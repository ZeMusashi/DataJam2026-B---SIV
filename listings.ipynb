```python
#data searching

import pandas as pd

listings = pd.read_csv("listings.csv.gz")

search = "Kitsilano"
column_name = "neighbourhood_cleansed"

count = 0

for neighbourhood in listings[column_name]:
    if neighbourhood.lower() == search.lower():
        count += 1

print(f"{search}: {count} found")
```

```python
#Price Calculations


import pandas as pd

listings = pd.read_csv("listings.csv.gz")

column_name = "neighbourhood_cleansed"
search = "Mount Pleasant"

# Clean the price column
listings["price"] = pd.to_numeric(
    listings["price"].astype("string").str.replace(r"[$,]", "", regex=True),
    errors="coerce"
)

total = 0
count = 0

for index, neighbourhood in enumerate(listings[column_name]):

    if neighbourhood == search:
        price = listings["price"].iloc[index]

        if pd.notna(price):
            total += price
            count += 1

if count == 0:
    print("No valid listings found.")

else:
    print(f"The average price is ${total / count:.2f}")
    print(f"There are a total of {count} listed prices")
```

```python
#Review Calculations


import pandas as pd

listings = pd.read_csv("listings.csv.gz")

column_name = "review_scores_value"
search = "Mount Pleasant"

total = 0
count = 0

for index, neighbourhood in enumerate(listings["neighbourhood_cleansed"]):

    if neighbourhood == search:
        score = listings[column_name].iloc[index]

        if pd.notna(score):
            total += score
            count += 1

if count == 0:
    print("No valid listings found.")

else:
    print(f"The average review score is {total / count:.2f}")
    print(f"There are a total of {count} reviews")
    
```
